```python
import nltk
import os

# Store datasets in the current project folder
nltk_data_path = os.path.abspath("nltk_data")
os.makedirs(nltk_data_path, exist_ok=True)

nltk.data.path.insert(0, nltk_data_path)

nltk.download('gutenberg', download_dir=nltk_data_path)
nltk.download('punkt', download_dir=nltk_data_path)
nltk.download('punkt_tab', download_dir=nltk_data_path)
nltk.download('stopwords', download_dir=nltk_data_path)

print("NLTK datasets downloaded!")
```

```python
from nltk.corpus import gutenberg, stopwords

print("Alice dataset:", len(gutenberg.raw('carroll-alice.txt')))
print("English stopwords:", len(stopwords.words('english')))
```

```python
nltk.data.path.insert(0, os.path.abspath("nltk_data"))

def check_nltk_data(path):
    try:
        nltk.data.find(path)
        return True
    except LookupError:
        return False

missing = []

if not check_nltk_data("corpora/gutenberg/carroll-alice.txt"):
    missing.append("gutenberg")

if not (
    check_nltk_data("tokenizers/punkt_tab")
    or check_nltk_data("tokenizers/punkt")
):
    missing.append("punkt_tab")

if not check_nltk_data("corpora/stopwords"):
    missing.append("stopwords")

if missing:
    print("Missing NLTK datasets:", missing)
    raise LookupError(
        "Please install/download the missing NLTK datasets."
    )
else:
    print("All NLTK datasets are available!")
```

```python
text = gutenberg.raw("carroll-alice.txt")

print("Dataset loaded successfully!")
print("Characters in raw text:", len(text))

print("\nSample text:")
print(text[:500])
```

```python
import nltk
from nltk.tokenize import sent_tokenize, word_tokenize
from nltk.corpus import stopwords, gutenberg
import string
import re
```

```python
sentences = sent_tokenize(text)

stop_words = set(stopwords.words("english"))

processed_sentences = []

for sentence in sentences:
    sentence = sentence.lower()
    sentence = re.sub(r"\d+", "", sentence)
    sentence = sentence.translate(str.maketrans("", "", string.punctuation))

    words = word_tokenize(sentence)

    words = [
        word for word in words
        if word not in stop_words and len(word) > 1
    ]

    if words:
        processed_sentences.append(words)

print("Total processed sentences:", len(processed_sentences))
print("First 5 sentences:")
print(processed_sentences[:5])
```
