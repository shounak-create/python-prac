## Step 2 - MNIST (Assignments 2, 3, 4, 6-ResNet)

```python
import os, urllib.request

def fetch(url, dest):
    if os.path.exists(dest):
        print("already have", dest); return
    print("downloading", dest, "...")
    urllib.request.urlretrieve(url, dest)
    print("  done:", round(os.path.getsize(dest)/1e6, 1), "MB")

fetch("https://storage.googleapis.com/tensorflow/tf-keras-datasets/mnist.npz", "mnist.npz")
```

## Step 3 - ResNet50 ImageNet weights (Assignment 6, ResNet50)

```python
fetch("https://storage.googleapis.com/tensorflow/keras-applications/resnet/"
      "resnet50_weights_tf_dim_ordering_tf_kernels_notop.h5",
      "resnet50_weights_tf_dim_ordering_tf_kernels_notop.h5")
```

## Step 4 - NLTK data (Assignment 5)

```python
import nltk
for pkg in ["gutenberg", "punkt", "punkt_tab", "stopwords"]:
    nltk.download(pkg, download_dir="nltk_data", quiet=True)
print(sorted(os.listdir("nltk_data")))
```

## Step 5 - YOLOv8n weights, COCO128 dataset and plot font (Assignment 6, object detection)

```python
import zipfile
fetch("https://github.com/ultralytics/assets/releases/download/v8.4.0/yolov8n.pt", "yolov8n.pt")

os.makedirs("datasets", exist_ok=True)
if not os.path.exists("datasets/coco128/images/train2017"):
    fetch("https://github.com/ultralytics/assets/releases/download/v0.0.0/coco128.zip", "datasets/coco128.zip")
    with zipfile.ZipFile("datasets/coco128.zip") as z:
        z.extractall("datasets")
print("coco128 images:", len(os.listdir("datasets/coco128/images/train2017")))

# Arial.ttf is used by Ultralytics for plots; store it where Ultralytics looks for it
from ultralytics.utils import USER_CONFIG_DIR
USER_CONFIG_DIR.mkdir(parents=True, exist_ok=True)
fetch("https://github.com/ultralytics/assets/releases/download/v0.0.0/Arial.ttf", str(USER_CONFIG_DIR / "Arial.ttf"))
```

## Step 6 - Check everything is in place

```python
needed = ["mnist.npz", "resnet50_weights_tf_dim_ordering_tf_kernels_notop.h5", "yolov8n.pt",
          "nltk_data", "datasets/coco128/images/train2017", "datasets/coco128/labels/train2017"]
for p in needed:
    print(("OK      " if os.path.exists(p) else "MISSING ") + p)
```
